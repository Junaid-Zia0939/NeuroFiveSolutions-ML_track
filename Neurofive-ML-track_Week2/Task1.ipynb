```python
House=House[['price','area','bathrooms','airconditioning']].copy()
```

```python
House.head()
```

```python
from sklearn.model_selection import train_test_split
x=House.drop(['price'],axis=1)
y=House['price']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.1,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import r2_score


# 1. R-Squared Score (How well the model fits the data: 1.0 is perfect)
r2 = r2_score(y_test, y_pred)
print(r2)
```

The R^2 score means that almost 56 percent of the house's price can be predicted accurately by these three features including area,bathrooms and airconditioning, while the rest of 44 percent of the price is determined by other metrics. It simply means that now our model can accurately predict the price of the house based on the three metrics upto 56 percent accuracy. We can improve our model prediction accuracy by training it with higher number of features.
