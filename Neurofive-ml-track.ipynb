```python
plt.figure(figsize=(10,6))

import numpy as np
numeric_data=titanic.select_dtypes(include=[np.number])
correlation_matrix=numeric_data.corr(method='pearson')
correlation_matrix
```

```python
sns.heatmap(correlation_matrix)
plt.title('Correlation Heatmap',fontsize=16, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
```

The correlation matrix and the heatmap clearly showed the column Fare has a higher impact on the target value i-e Survived. This is because those who paid the higher fare was given cabins in the first class i-e 1 class. It means there were luxurious and rich people travelling in this class, and thus when the incident occurred, more rescue operations were launched and hurried to save the people in the 1 (first) class, and thus more people were saved from drowning in the first class while people in the other lower classes died the most. Thus, fare mostly determines the target survivied.
